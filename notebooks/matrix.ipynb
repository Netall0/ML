# with Bias

```python
import numpy as np


X = np.array([[22,7],
              [35,50],
              [18,12]]) # (3*2)

Y = np.array([[1],
              [3],
              ]) #(2*1)

b = -1.2



z = X @ Y + b

print(z)
```

```python
X_original = np.array([[22, 7],
                       [35, 50],
                       [18, 12]])

X_with_bias = np.hstack([np.ones((X_original.shape[0], 1)), X_original])

print(X_with_bias)

W_combined = np.array([[-1.2], 
                       [ 0.1],  
                       [ 0.5]]) 




z = X_with_bias @ W_combined
print(z)
```
