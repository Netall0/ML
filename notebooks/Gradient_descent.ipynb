```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def objective_function(x):
    return x**2
```

```python
objective_function(2)
```

```python
bounds = np.array([[-10,10]]) # create a 2D array
print(bounds)

```

```python
from matplotlib.pylab import rand


x = bounds[:,0] + rand(len(bounds)) * (bounds[:,1] - bounds[:,0])
print(x)
```

```python
data = []
target = []

for i in range (1000):
    x = bounds[:,0] + rand(len(bounds)) * (bounds[:,1] - bounds[:,0])
    data.append(x)
    target.append(objective_function(x))

plt.scatter(data,target)
```

```python
x = bounds[:,0] + rand(len(bounds)) * (bounds[:,1] - bounds[:,0])
x
```

```python
import  sympy  as sp


xs = sp.Symbol('x')
obj_func = xs**2

def get_grad(val):
    df = sp.diff(obj_func, xs)
    return float(df.subs(xs, val))

```

```python
get_grad(float(x))
```

```python
new_x = x - 0.1 * get_grad(float(x))
new_x
```

```python
new_x = x - 0.1 * get_grad(float(x))
new_x
```

```python
solutions = []
evalutions = []
```

```python
for i in range(20):
    solutions.append(x)
    x_eval = objective_function(x)
    evalutions.append(x_eval)
    
    gradient = get_grad(float(x))
    new_x = x - 0.1 * gradient
    x = new_x


plt.scatter(data,target)
plt.scatter(solutions,evalutions)
```
